***2.Identify the shapes of arrays a and b before addition.***

a has 3 rows and 1 column

***3.Explain how NumPy broadcasting enables the addition of these arrays.***

NumPy automatically expands the smaller dimensions to make the shapes compatible.
Conceptually:

a:                 b:

10                 1  2  3

20       +        1  2  3

30                1  2  3

So the addition becomes:

10+1   10+2   10+3

20+1   20+2   20+3

30+1   30+2   30+3

Result:

[[11 12 13]

 [21 22 23]

 [31 32 33]]

***4.What would happen if b were replaced with np.array([1, 2])? Explain the reason.***

NumPy broadcasts b from (2,) to (1, 2) and a from (3,1) across the columns.

NumPy compares dimensions from right to left.b has only 2 elements, but because a has 1 column, NumPy can broadcast that single column to match the 2 elements of b. Therefore, the addition is valid and produces a (3, 2) array.

# Q1. A student assessment dataset contains missing marks, duplicate student records, inconsistent department names such as IT, it, and I.T., and marks stored as text. Develop a cleaning sequence to standardize the data, handle missing values, remove duplicates, and convert marks into a suitable numeric type. State one risk of deleting missing values without investigating why they are missing.

Data cleaning is the process of correcting missing, duplicate, inconsistent, and incorrectly formatted data. In this dataset, first we investigate and handle missing marks such as blank values, NA, and Absent. Next, duplicate student records are removed using Student_ID. Department names such as IT, it, and I.T. are standardized into a common format. The Marks column is cleaned by removing unnecessary text such as "marks" and converting the values into a numeric type. Finally, the cleaned data should be validated before analysis. Missing values should not be deleted without investigation because they may have meaningful reasons, such as a student being absent, and deleting them blindly can cause information loss and biased results.




```python
import pandas as pd

df = pd.read_csv(r"C:\Users\DELL\Desktop\assignment\student_assessment_dirty.csv")

# 1. Standardize department names
df["Department"] = df["Department"].str.upper()
df["Department"] = df["Department"].str.replace(".", "", regex=False)




# 3. Remove duplicate student records
df = df.drop_duplicates(subset="Student_ID")

# 4. Remove text from marks
df["Marks"] = df["Marks"].str.replace(" marks", "", regex=False)

# 5. Convert Marks into numeric type
df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")

# 6. Handle missing marks using the mean
mean_marks = df["Marks"].mean()
df["Marks"] = df["Marks"].fillna(mean_marks)
print(df)
print(df.dtypes)
```

```python
import pandas as pd

# Load the CSV file
df = pd.read_csv(r"C:\Users\DELL\Desktop\assignment\student_assessment_dirty.csv")

# Display first 5 rows
display(df.head())

# Display dataset information
display( df.info())

# Check missing values in each column
display( df.isnull().sum())
```

***2.Standardize department names so that IT, it, and I.T. are represented consistently. Apply the same approach to Finance.***

.str.upper() handles differences like it → IT and finance → FINANCE, while .replace() handles special formats such as I.T..

```python
# Standardize department names
df["Department"] = (
    df["Department"]
    .str.strip()
    .str.upper()
    .replace({
        "I.T.": "IT",
        "IT": "IT",
        "FINANCE": "FINANCE",
        "FIN.": "FINANCE"
    })
)

print(df["Department"].unique())
```

***3.Identify and remove exact duplicate student assessment records.***

df.duplicated().sum() → counts the exact duplicate rows.

df[df.duplicated()] → displays the duplicate records.

df.drop_duplicates() → removes the duplicate rows.

The final check confirms that no exact duplicates remain.


```python
# Identify duplicate records
print("Number of duplicate records:", df.duplicated().sum())

# Display duplicate records
print(df[df.duplicated()])

# Remove exact duplicate records
df = df.drop_duplicates()

# Verify that duplicates are removed
print("Duplicates after removal:", df.duplicated().sum())
```

***4.Convert the Marks column into a numeric format, handling "88 marks", "Absent", "NA", and blank entries appropriately.***

Here, errors="coerce" converts invalid values such as Absent, NA, and blanks into NaN, which can then be handled using the mean in the next step.


```python
# Convert Marks column to numeric
df["Marks"] = (
    df["Marks"]
    .astype(str)
    .str.replace("marks", "", case=False, regex=False)
    .str.strip()
)

df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")

print(df["Marks"])
```

***5.Investigate missing marks and distinguish absent students from students whose marks have not been entered. State one risk of deleting missing values without investigating why they are missing.***

Deleting all missing-mark records without investigating can remove absent students or a particular group of students, which may lead to biased analysis and incorrect conclusions about student performance.

"Absent" → should be identified separately rather than treating it as an ordinary missing mark.
"NA" or blank → investigate whether the mark was simply not entered.
Only after investigation should we decide how to handle the missing values.


```python
# Check records where Marks are missing
missing_marks = df[df["Marks"].isnull()]

print(missing_marks)
```

***6.Identify marks outside the valid range of 0–100 and decide how they should be handled***

First, find marks that are less than 0 or greater than 100

If invalid values are found, convert them to NaN so they can be investigated or handled appropriately but there are no invalid values in this dataset

```python
# Identify invalid marks
invalid_marks = df[(df["Marks"] < 0) | (df["Marks"] > 100)]

print("Invalid marks:")
print(invalid_marks)
```

***7.Display the cleaned DataFrame and report the number of records removed, missing marks remaining, and invalid marks identified***

```python
# Number of records before removing duplicates
records_before = len(df)

# Remove exact duplicates
df = df.drop_duplicates()

# Number of records removed
records_removed = records_before - len(df)

# Identify invalid marks (outside 0–100)
invalid_marks = df[(df["Marks"] < 0) | (df["Marks"] > 100)]

# Count invalid marks
invalid_count = len(invalid_marks)

# Handle invalid marks by converting them to NaN
df.loc[(df["Marks"] < 0) | (df["Marks"] > 100), "Marks"] = pd.NA

# Handle missing marks using mean
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

# Count missing marks remaining
missing_remaining = df["Marks"].isnull().sum()

# Display cleaned DataFrame
print("Cleaned DataFrame:")
print(df)

# Report cleaning results
print("\n--- Cleaning Report ---")
print("Records removed:", records_removed)
print("Missing marks remaining:", missing_remaining)
print("Invalid marks identified:", invalid_count)
```
